```python

num1 =int(input("Enter 1st number: "))
num2 =int(input("Enter 2nd number: "))
operator=input("Enter the operator (+, -, *, /): ")

if operator=='+':
    res=num1+num2
elif operator=='-':
    res=num1-num2
elif operator=='*':
    res=num1*num2
elif operator=='/':
    res=num1/num2
else:
    res="wrong input"

print("The result is", res)
```
